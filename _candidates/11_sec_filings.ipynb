# QuantJourney SDK - SEC Company Filings

This notebook demonstrates SEC data:
- Company filings (10-K, 10-Q, 8-K)
- Regulatory documents
- Corporate disclosures

**API:** https://api.quantjourney.cloud

```python
import sys
sys.path.insert(0, '..')
from quantjourney.sdk import QuantJourneyAPI
import pandas as pd
import os
API_KEY = os.environ['QJ_API_KEY']
qj = QuantJourneyAPI(api_key=API_KEY)
print('✓ Connected to QuantJourney API')
```

## 1. 10-K Annual Reports

```python
try:
    response = qj.sec.get_filings(symbol='AAPL', filing_type='10-K', limit=10)
    filings = response.get('value', response) if isinstance(response, dict) else response
    if filings:
        print('AAPL - Annual Reports (10-K)')
        print('=' * 60)
        for i, f in enumerate(filings[:5], 1):
            if isinstance(f, dict):
                date = f.get('filingDate', f.get('date', 'N/A'))
                form = f.get('form', f.get('type', '10-K'))
                print(f'\n{i}. Filing Date: {date}')
                print(f'   Form: {form}')
except Exception as e:
    print(f'Note: SEC 10-K filings - {e}')
```

## 2. SEC Filing Types Reference

```python
print('=' * 60)
print('SEC FILING TYPES REFERENCE')
print('=' * 60)
filing_types = [('10-K', 'Annual report - comprehensive business overview'), ('10-Q', 'Quarterly report - unaudited financial statements'), ('8-K', 'Current report - material events disclosure'), ('DEF 14A', 'Proxy statement - shareholder meeting info'), ('S-1', 'Registration statement - IPO documentation'), ('4', 'Insider trading - officer/director transactions'), ('13F', 'Institutional holdings - quarterly positions')]
print(f"\n{'Form':<12} {'Description'}")
print('-' * 55)
for form, desc in filing_types:
    print(f'{form:<12} {desc}')
```

# QuantJourney SDK - SEC Company Filings

This notebook demonstrates SEC data:
- Company filings (10-K, 10-Q, 8-K)
- Regulatory documents
- Corporate disclosures

**API:** https://api.quantjourney.cloud

```python
import sys
sys.path.insert(0, '..')
from quantjourney.sdk import QuantJourneyAPI
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
pio.renderers.default = 'notebook+png'
import plotly.graph_objects as go
import os
API_KEY = os.environ['QJ_API_KEY']
qj = QuantJourneyAPI(api_key=API_KEY)
print('✓ Connected to QuantJourney API')
```

## 1. 10-K Annual Reports

```python
try:
    response = qj.sec.get_filings(symbol='AAPL', filing_type='10-K', limit=10)
    filings = response.get('value', response) if isinstance(response, dict) else response
    if filings:
        print('AAPL - Annual Reports (10-K)')
        print('=' * 60)
        for i, f in enumerate(filings[:5], 1):
            if isinstance(f, dict):
                date = f.get('filingDate', f.get('date', 'N/A'))
                form = f.get('form', f.get('type', '10-K'))
                desc = f.get('description', '')[:50]
                url = f.get('url', f.get('link', 'N/A'))
                print(f'\n{i}. Filing Date: {date}')
                print(f'   Form: {form}')
                if desc:
                    print(f'   Description: {desc}...')
except Exception as e:
    print(f'Note: SEC 10-K filings - {e}')
```

## 2. 10-Q Quarterly Reports

```python
try:
    response = qj.sec.get_filings(symbol='MSFT', filing_type='10-Q', limit=10)
    filings_q = response.get('value', response) if isinstance(response, dict) else response
    if filings_q:
        print('MSFT - Quarterly Reports (10-Q)')
        print('=' * 60)
        for i, f in enumerate(filings_q[:5], 1):
            if isinstance(f, dict):
                date = f.get('filingDate', f.get('date', 'N/A'))
                form = f.get('form', f.get('type', '10-Q'))
                print(f'\n{i}. Filing Date: {date}')
                print(f'   Form: {form}')
except Exception as e:
    print(f'Note: SEC 10-Q filings - {e}')
```

## 3. 8-K Current Reports

```python
try:
    response = qj.sec.get_filings(symbol='TSLA', filing_type='8-K', limit=10)
    filings_8k = response.get('value', response) if isinstance(response, dict) else response
    if filings_8k:
        print('TSLA - Current Reports (8-K)')
        print('=' * 60)
        print('\n8-K filings report material events like:')
        print('  - Earnings releases')
        print('  - Executive changes')
        print('  - Mergers & acquisitions')
        print('  - Bankruptcy')
        print('\nRecent 8-K filings:')
        for i, f in enumerate(filings_8k[:5], 1):
            if isinstance(f, dict):
                date = f.get('filingDate', f.get('date', 'N/A'))
                desc = f.get('description', f.get('items', 'N/A'))
                print(f'\n{i}. Date: {date}')
                if desc:
                    print(f'   Items: {str(desc)[:60]}...')
except Exception as e:
    print(f'Note: SEC 8-K filings - {e}')
```

## 4. Multiple Companies Comparison

```python
symbols = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'NVDA']
filing_counts = {}
print('Recent SEC Filings by Company')
print('=' * 60)
for symbol in symbols:
    try:
        response = qj.sec.get_filings(symbol=symbol, limit=20)
        filings = response.get('value', response) if isinstance(response, dict) else response
        if filings:
            count = len(filings)
            filing_counts[symbol] = count
            latest = filings[0] if filings else {}
            latest_date = latest.get('filingDate', latest.get('date', 'N/A')) if isinstance(latest, dict) else 'N/A'
            latest_type = latest.get('form', latest.get('type', 'N/A')) if isinstance(latest, dict) else 'N/A'
            print(f'\n{symbol}:')
            print(f'  Recent filings: {count}')
            print(f'  Latest: {latest_type} on {latest_date}')
    except Exception as e:
        print(f'\n{symbol}: Error - {e}')
```

## 5. SEC Filing Types Reference

```python
print('=' * 60)
print('SEC FILING TYPES REFERENCE')
print('=' * 60)
filing_types = [('10-K', 'Annual report - comprehensive business overview'), ('10-Q', 'Quarterly report - unaudited financial statements'), ('8-K', 'Current report - material events disclosure'), ('DEF 14A', 'Proxy statement - shareholder meeting info'), ('S-1', 'Registration statement - IPO documentation'), ('4', 'Insider trading - officer/director transactions'), ('13F', 'Institutional holdings - quarterly positions'), ('13D/13G', 'Beneficial ownership - 5%+ stakes'), ('6-K', 'Foreign issuer reports'), ('20-F', 'Foreign company annual report')]
print(f"\n{'Form':<12} {'Description'}")
print('-' * 55)
for form, desc in filing_types:
    print(f'{form:<12} {desc}')
print('\n' + '=' * 60)
```
