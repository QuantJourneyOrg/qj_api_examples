# QuantJourney SDK - OpenFIGI Instrument Identification

This notebook demonstrates OpenFIGI data:
- Global instrument identification
- FIGI lookup
- Cross-market symbol mapping

**API:** https://api.quantjourney.cloud

```python
# Setup
import sys
sys.path.insert(0, '..')

from quantjourney.sdk import QuantJourneyAPI
import pandas as pd
import numpy as np

# API Key authentication
import os
API_KEY = os.environ.get("QJ_API_KEY", "qj_...")
qj = QuantJourneyAPI(api_key=API_KEY)
print("✓ Connected to QuantJourney API")
```

## 1. Search by Company Name

```python
# Search OpenFIGI by company name
try:
    response = qj.openfigi.search(query="Apple Inc")
    figi_data = response.get('value', response) if isinstance(response, dict) else response
    
    if figi_data:
        print("OpenFIGI Search: 'Apple Inc'")
        print("=" * 60)
        
        results = figi_data[:10] if isinstance(figi_data, list) else [figi_data]
        for i, item in enumerate(results, 1):
            if isinstance(item, dict):
                print(f"\n{i}. {item.get('name', 'N/A')}")
                print(f"   FIGI:     {item.get('figi', 'N/A')}")
                print(f"   Ticker:   {item.get('ticker', 'N/A')}")
                print(f"   Exchange: {item.get('exchCode', item.get('exchange', 'N/A'))}")
                print(f"   Type:     {item.get('securityType', item.get('type', 'N/A'))}")
except Exception as e:
    print(f"Note: OpenFIGI search - {e}")
```

## 2. Lookup by Ticker

```python
# Search by ticker
tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'NVDA']

print("OpenFIGI Lookup by Ticker")
print("=" * 60)

for ticker in tickers:
    try:
        response = qj.openfigi.search(query=ticker)
        data = response.get('value', response) if isinstance(response, dict) else response
        
        if data:
            item = data[0] if isinstance(data, list) else data
            if isinstance(item, dict):
                figi = item.get('figi', 'N/A')
                name = item.get('name', 'N/A')[:30]
                print(f"{ticker:>6} | FIGI: {figi:<15} | {name}")
    except Exception as e:
        print(f"{ticker:>6} | Error: {e}")
```

## 3. Cross-Market Mapping

```python
# Find same security across different exchanges
try:
    response = qj.openfigi.search(query="Microsoft")
    msft_data = response.get('value', response) if isinstance(response, dict) else response
    
    if msft_data:
        print("Microsoft - Cross-Market Listings")
        print("=" * 60)
        
        results = msft_data[:15] if isinstance(msft_data, list) else [msft_data]
        
        # Group by exchange
        exchanges = {}
        for item in results:
            if isinstance(item, dict):
                exch = item.get('exchCode', item.get('exchange', 'Unknown'))
                if exch not in exchanges:
                    exchanges[exch] = []
                exchanges[exch].append(item)
        
        for exch, items in exchanges.items():
            print(f"\n{exch}:")
            for item in items[:3]:
                ticker = item.get('ticker', 'N/A')
                figi = item.get('figi', 'N/A')
                sec_type = item.get('securityType', 'N/A')
                print(f"  {ticker:>10} | {figi} | {sec_type}")
except Exception as e:
    print(f"Note: Cross-market mapping - {e}")
```
